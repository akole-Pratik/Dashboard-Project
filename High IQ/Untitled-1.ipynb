```python
# Define custom age groups
age_groups = {
    '18-25': range(18, 26),
    '26-35': range(26, 36),
    '36-45': range(36, 46),
    '46-55': range(46, 56),
    '56-65': range(56, 66),
}

# Assign age groups manually
def assign_age_group(age):
    for group, age_range in age_groups.items():
        if age in age_range:
            return group
    return '60+'

df['Age Group'] = df['Age'].apply(assign_age_group)

# Group by age groups and categories
grouped_data = df.groupby(['Age Group', 'Category']).size().reset_index(name='Count')

print(grouped_data)
```

```python
import pandas as pd

data = {'Year': [2022, 2022, 2023, 2023, 2024],
        'Quarter': ['Q1', 'Q2', 'Q1', 'Q2','Q3'],
        'Sales': [100, 120, 150, 180, 210]}
df = pd.DataFrame(data)
print(df)
```

```python
df_unstacked = df.set_index(['Year', 'Quarter']).unstack('Quarter', fill_value=0)
print(df_unstacked)
```
